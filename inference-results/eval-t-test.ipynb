```python
import evaluate
from sentence_transformers import SentenceTransformer
from tqdm import tqdm

bleu = evaluate.load("bleu")
rouge = evaluate.load('rouge')

model = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

def compute_score(generated_list_of_content, reference):
    bleu_score = bleu.compute(predictions=generated_list_of_content, references=reference)
    precision_keys = ['bleu1', 'bleu2', 'bleu3', 'bleu4']
    for i, key in enumerate(precision_keys):
        bleu_score[key] = bleu_score['precisions'][i]    
    del bleu_score['precisions']
    rouge_score = rouge.compute(predictions=generated_list_of_content, references=reference)

    bert_scores = [model.similarity(model.encode(each), model.encode(reference[i]))[0] 
                                    for i, each in 
                                        enumerate(generated_list_of_content)
                                        ]
    average_bert_scores = {"bert_score": sum(bert_scores)/len(bert_scores)}

    return {**bleu_score, 
            **{k: float(v) for k, v in rouge_score.items()}, 
            **{k: float(v) for k, v in average_bert_scores.items()}}
```

```python
all_scores = []
for col in baseline_squad.columns[1:]:
    temp_scores = []
    for i, item in enumerate(tqdm(baseline_squad['reference'])):
        scores = compute_score([baseline_squad.loc[i, col]], [item])
        temp_scores.append(scores)
    all_scores.append({'column': col, 'scores': temp_scores})
```

```python
import pandas as pd
import numpy as np
from scipy import stats

def run_ttests(all_scores):
    """
    Run paired t-tests across models for multiple score metrics.
    all_scores[0] is treated as baseline.
    """
    
    def significance_label(p):
        return "significant" if p < 0.05 else "not significant"

    def cohens_d_paired(x, y):
        diff = np.array(x) - np.array(y)
        return diff.mean() / diff.std(ddof=1)

    score_list = ['bleu1', 'bleu2', 'bleu3', 'bleu4',
                  'rouge1', 'rouge2', 'rougeL', 'bert_score']

    rows = []

    baseline = all_scores[0]
    baseline_label = baseline['column']

    for score_name in score_list:
        baseline_scores = [s[score_name] for s in baseline['scores']]

        for item in all_scores[1:]:
            current_scores = [s[score_name] for s in item['scores']]

            # paired differences
            diffs = np.array(baseline_scores) - np.array(current_scores)
            sd_diff = diffs.std(ddof=1)

            # paired t-test
            t_stat, p_value = stats.ttest_rel(baseline_scores, current_scores)

            # Cohen’s d
            d = cohens_d_paired(baseline_scores, current_scores)

            rows.append({
                "score": score_name,
                "baseline": baseline_label,
                "comparison": item['column'],
                "cohens_d": d,
                "sd_diff": sd_diff,
                "t_stat": t_stat,
                "p_value": p_value,
                "significance": significance_label(p_value),
            })

    return pd.DataFrame(rows)
```

```python
df_ttest = run_ttests(all_scores)
df_ttest
```
