```python
target_columns = sorted([item for item in grouped_squad_eval_on_context.columns if '-' in item]) 
target_columns_tydiqa = sorted([item for item in grouped_tydiqa_eval_on_context.columns if '-' in item])
target_columns
```

```python
from typing import List
import numpy as np
from sentence_transformers import SentenceTransformer, util
from tqdm import tqdm

# Load model once globally for efficiency
bert_model = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')

def ngitung(predicted: List[str], actual: List[str]):
    """
    Compute QA-style Exact Match (EM), Precision, Recall, F1, number of predictions,
    and BERT semantic similarity score.
    """
    # Normalize
    predicted = [p.lower().strip() for p in predicted]
    actual = [a.lower().strip() for a in actual]

    # Pad shorter list
    max_len = max(len(predicted), len(actual))
    preds = predicted + [''] * (max_len - len(predicted))
    refs = actual + [''] * (max_len - len(actual))

    em_scores, f1_scores, precision_scores, recall_scores, bert_sims = [], [], [], [], []

    # Compute embeddings once
    embeddings_pred = bert_model.encode(preds, convert_to_tensor=True)
    embeddings_ref = bert_model.encode(refs, convert_to_tensor=True)

    for i, (p, r) in tqdm(enumerate(zip(preds, refs)), total=len(refs)):
        # Exact Match
        em = int(p == r)
        em_scores.append(em)

        # Token-level metrics
        p_tokens = p.split()
        r_tokens = r.split()
        common_tokens = set(p_tokens) & set(r_tokens)
        if not common_tokens:
            precision = recall = f1 = 0
        else:
            precision = len(common_tokens) / len(p_tokens) if p_tokens else 0
            recall = len(common_tokens) / len(r_tokens) if r_tokens else 0
            f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0
        precision_scores.append(precision)
        recall_scores.append(recall)
        f1_scores.append(f1)

        # BERT similarity (cosine)
        bert_sim = util.cos_sim(embeddings_pred[i], embeddings_ref[i]).item()
        bert_sims.append(bert_sim)

    return {
        'num_predictions': len(predicted),
        'exact_match': np.mean(em_scores) ,
        'precision': np.mean(precision_scores) ,
        'recall': np.mean(recall_scores) ,
        'f1': np.mean(f1_scores) ,
        'bert_similarity': np.mean(bert_sims)   
    }
```

```python
import pandas as pd

# Initialize a list to hold the results
results_list = []

for col in target_columns:
    predicted_list = grouped_squad_eval_on_context[col].explode().tolist()
    actual_list = grouped_squad_eval_on_context['answer'].explode().tolist()
    
    scores = ngitung(predicted=predicted_list, actual=actual_list)
    
    # Append results as a dict
    results_list.append({
        'column': col,
        'num_predictions': scores['num_predictions'],
        'exact_match': scores['exact_match'],
        'precision': scores['precision'],
        'recall': scores['recall'],
        'f1': scores['f1'],
        'bert_score': scores['bert_similarity']
    })

# Convert list of dicts into a DataFrame
scores_df = pd.DataFrame(results_list)

# Optional: format the numbers to 2 decimal places
scores_df[['exact_match', 'precision', 'recall', 'f1', 'bert_score']] = scores_df[['exact_match', 'precision', 'recall', 'f1', 'bert_score']].round(4)
```

```python
import pandas as pd

# Initialize a list to hold the results
results_list = []

for col in target_columns_tydiqa:
    predicted_list = grouped_tydiqa_eval_on_context[col].explode().tolist()
    actual_list = grouped_tydiqa_eval_on_context['answer'].explode().tolist()
    
    scores = ngitung(predicted=predicted_list, actual=actual_list)
    
    # Append results as a dict
    results_list.append({
        'column': col,
        'num_predictions': scores['num_predictions'],
        'exact_match': scores['exact_match'],
        'precision': scores['precision'],
        'recall': scores['recall'],
        'f1': scores['f1'],
        'bert_score': scores['bert_similarity']
    })

# Convert list of dicts into a DataFrame
scores_df_tydiqa = pd.DataFrame(results_list)

# Optional: format the numbers to 2 decimal places
scores_df_tydiqa[['exact_match', 'precision', 'recall', 'f1', 'bert_score']] = scores_df_tydiqa[['exact_match', 'precision', 'recall', 'f1', 'bert_score']].round(4)
```
